```python
def segment_analysis(data, group_col):
    result = (
        data.groupby(group_col, observed=True)
            .agg(
                PolicyCount=("IDpol", "nunique"),
                Exposure=("Exposure", "sum"),
                ClaimCount=("ClaimNb", "sum"),
                TotalClaimCost=("TotalClaimCost", "sum")
            )
            .reset_index()
    )

    result["ClaimFrequency"] = (
        result["ClaimCount"] /
        result["Exposure"]
    )

    result["AverageSeverity"] = np.where(
        result["ClaimCount"] > 0,
        result["TotalClaimCost"] /
        result["ClaimCount"],
        0
    )

    return result
```
